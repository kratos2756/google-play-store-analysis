```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/kratos2756/google-play-store-analysis/refs/heads/main/dataset/googleplaystore.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

display(df.head())
```

```python
print("Columns in the dataset:")
print(df.columns.tolist())
```

```python
df.info()
```

```python
# Check missing values in each column

missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
# Check duplicate rows

duplicate_rows = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_rows)
```

```python
# Check duplicate application names

duplicate_apps = df['App'].duplicated().sum()

print("Number of duplicate app names:", duplicate_apps)
```

```python
# Display duplicate application records

duplicate_data = df[df['App'].duplicated(keep=False)].sort_values('App')

display(duplicate_data.head(20))
```

```python
# Remove duplicate application records

df = df.drop_duplicates(subset='App', keep='first')

print("Dataset shape after removing duplicate apps:")
print(df.shape)
```

```python
# Check missing values after duplicate removal

print(df.isnull().sum())
```

```python
print(df['Rating'].describe())
print("Missing ratings:", df['Rating'].isnull().sum())
```

```python
# Fill missing ratings with the median rating

median_rating = df['Rating'].median()

df['Rating'] = df['Rating'].fillna(median_rating)

print("Missing ratings after cleaning:", df['Rating'].isnull().sum())
```

```python
print(df['Reviews'].dtype)
```

```python
# Convert Reviews to numeric

df['Reviews'] = pd.to_numeric(df['Reviews'], errors='coerce')

print(df['Reviews'].dtype)
```

```python
# Clean the Installs column

df['Installs'] = (
    df['Installs']
    .astype(str)
    .str.replace(',', '', regex=False)
    .str.replace('+', '', regex=False)
)

df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce')

print(df['Installs'].head())
print(df['Installs'].dtype)
```

```python
# Clean the Price column

df['Price'] = (
    df['Price']
    .astype(str)
    .str.replace('$', '', regex=False)
)

df['Price'] = pd.to_numeric(df['Price'], errors='coerce')

print(df['Price'].head())
print(df['Price'].dtype)
```

```python
def convert_size(size):
    size = str(size).strip()

    if size == 'nan' or size == 'Varies with device':
        return np.nan

    if size.endswith('M'):
        return float(size[:-1])

    if size.endswith('k'):
        return float(size[:-1]) / 1024

    return np.nan
```

```python
df['Size_MB'] = df['Size'].apply(convert_size)

print(df[['Size', 'Size_MB']].head(10))
```

```python
df['Last Updated'] = pd.to_datetime(
    df['Last Updated'],
    errors='coerce'
)

print(df['Last Updated'].dtype)
```
