```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# GitHub Raw URL of cleaned dataset
url = "https://raw.githubusercontent.com/kratos2756/google-play-store-analysis/refs/heads/main/dataset/cleaned_data.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

display(df.head())
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
# Count number of apps in each category

category_counts = df['Category'].value_counts()

print(category_counts)
```

```python
plt.figure(figsize=(12, 7))

category_counts.sort_values().plot(kind='barh')

plt.title('Number of Apps by Category', fontsize=16)
plt.xlabel('Number of Apps', fontsize=12)
plt.ylabel('Category', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 7))

category_counts.sort_values().plot(kind='barh')

plt.title('Number of Apps by Category', fontsize=16)
plt.xlabel('Number of Apps', fontsize=12)
plt.ylabel('Category', fontsize=12)

plt.tight_layout()

plt.savefig('01_apps_by_category.png', dpi=300, bbox_inches='tight')

plt.show()
```

```python
most_common_category = category_counts.idxmax()
highest_count = category_counts.max()

print(
    f"The category with the highest number of apps is "
    f"{most_common_category}, with {highest_count} apps."
)
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    df['Rating'].dropna(),
    bins=20,
    edgecolor='black'
)

plt.title('Distribution of App Ratings', fontsize=16)
plt.xlabel('Rating', fontsize=12)
plt.ylabel('Number of Apps', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    df['Rating'].dropna(),
    bins=20,
    edgecolor='black'
)

plt.title('Distribution of App Ratings', fontsize=16)
plt.xlabel('Rating', fontsize=12)
plt.ylabel('Number of Apps', fontsize=12)

plt.tight_layout()

plt.savefig(
    '02_rating_distribution.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
print("Average rating:", round(df['Rating'].mean(), 2))
print("Median rating:", round(df['Rating'].median(), 2))
print("Minimum rating:", df['Rating'].min())
print("Maximum rating:", df['Rating'].max())
```

```python
rating_counts = df['Rating'].round(1).value_counts().sort_index()

most_common_rating = rating_counts.idxmax()
most_common_rating_count = rating_counts.max()

print(
    f"The most frequently occurring rating is approximately "
    f"{most_common_rating}, with {most_common_rating_count} apps."
)
```

```python
# Count free and paid applications

type_counts = df['Type'].value_counts()

print(type_counts)
```

```python
plt.figure(figsize=(8, 8))

plt.pie(
    type_counts,
    labels=type_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Free vs Paid Applications', fontsize=16)

plt.show()
```

```python
plt.figure(figsize=(8, 8))

plt.pie(
    type_counts,
    labels=type_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Free vs Paid Applications', fontsize=16)

plt.tight_layout()

plt.savefig(
    '03_free_vs_paid.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
total_apps = type_counts.sum()

for app_type, count in type_counts.items():
    percentage = (count / total_apps) * 100
    print(f"{app_type}: {count} apps ({percentage:.2f}%)")
```

```python
most_common_type = type_counts.idxmax()
most_common_count = type_counts.max()

print(
    f"{most_common_type} applications are the majority, "
    f"with {most_common_count} apps."
)
```

```python
# Calculate average rating for each category

category_avg_rating = (
    df.groupby('Category')['Rating']
    .mean()
    .sort_values()
)

print(category_avg_rating)
```

```python
plt.figure(figsize=(12, 8))

category_avg_rating.plot(kind='barh')

plt.title('Average Rating by App Category', fontsize=16)
plt.xlabel('Average Rating', fontsize=12)
plt.ylabel('Category', fontsize=12)

plt.xlim(0, 5)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))

category_avg_rating.plot(kind='barh')

plt.title('Average Rating by App Category', fontsize=16)
plt.xlabel('Average Rating', fontsize=12)
plt.ylabel('Category', fontsize=12)

plt.xlim(0, 5)

plt.tight_layout()

plt.savefig(
    '04_category_average_rating.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
highest_category = category_avg_rating.idxmax()
highest_rating = category_avg_rating.max()

lowest_category = category_avg_rating.idxmin()
lowest_rating = category_avg_rating.min()

print(
    f"Highest average rating: {highest_category} "
    f"({highest_rating:.2f})"
)

print(
    f"Lowest average rating: {lowest_category} "
    f"({lowest_rating:.2f})"
)
```

```python
# Calculate total installs for each category

category_installs = (
    df.groupby('Category')['Installs']
    .sum()
    .sort_values()
)

print(category_installs)
```

```python
plt.figure(figsize=(12, 8))

category_installs.plot(kind='barh')

plt.title('Total Installs by App Category', fontsize=16)
plt.xlabel('Total Installs', fontsize=12)
plt.ylabel('Category', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))

plt.barh(
    category_installs.index,
    category_installs.values
)

plt.xscale('log')

plt.title('Total Installs by App Category', fontsize=16)
plt.xlabel('Total Installs (Log Scale)', fontsize=12)
plt.ylabel('Category', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))

plt.barh(
    category_installs.index,
    category_installs.values
)

plt.xscale('log')

plt.title('Total Installs by App Category', fontsize=16)
plt.xlabel('Total Installs (Log Scale)', fontsize=12)
plt.ylabel('Category', fontsize=12)

plt.tight_layout()

plt.savefig(
    '06_category_installs.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
highest_install_category = category_installs.idxmax()
highest_install_count = category_installs.max()

print(
    f"Category with the highest total installs: "
    f"{highest_install_category}"
)

print(
    f"Total installs: {highest_install_count:,.0f}"
)
```

```python
# Select only paid applications

paid_apps = df[df['Type'] == 'Paid'].copy()

print("Number of paid apps:", len(paid_apps))
display(paid_apps[['App', 'Price']].head(10))
```

```python
print(paid_apps['Price'].describe())
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    paid_apps['Price'].dropna(),
    bins=30,
    edgecolor='black'
)

plt.title('Price Distribution of Paid Applications', fontsize=16)
plt.xlabel('Price (USD)', fontsize=12)
plt.ylabel('Number of Paid Apps', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
paid_apps_filtered = paid_apps[
    paid_apps['Price'] <= 100
]

plt.figure(figsize=(10, 6))

plt.hist(
    paid_apps_filtered['Price'].dropna(),
    bins=30,
    edgecolor='black'
)

plt.title(
    'Price Distribution of Paid Applications (Up to $100)',
    fontsize=16
)

plt.xlabel('Price (USD)', fontsize=12)
plt.ylabel('Number of Paid Apps', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    paid_apps_filtered['Price'].dropna(),
    bins=30,
    edgecolor='black'
)

plt.title(
    'Price Distribution of Paid Applications (Up to $100)',
    fontsize=16
)

plt.xlabel('Price (USD)', fontsize=12)
plt.ylabel('Number of Paid Apps', fontsize=12)

plt.tight_layout()

plt.savefig(
    '07_price_distribution.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
print("Average paid app price: $", round(paid_apps['Price'].mean(), 2))
print("Median paid app price: $", round(paid_apps['Price'].median(), 2))
print("Minimum paid app price: $", round(paid_apps['Price'].min(), 2))
print("Maximum paid app price: $", round(paid_apps['Price'].max(), 2))
```

```python
size_install_data = df[['Size_MB', 'Installs']].dropna()

plt.figure(figsize=(10, 6))

plt.scatter(
    size_install_data['Size_MB'],
    size_install_data['Installs'],
    alpha=0.5
)

plt.yscale('log')

plt.title('App Size vs Installations', fontsize=16)
plt.xlabel('App Size (MB)', fontsize=12)
plt.ylabel('Number of Installs (Log Scale)', fontsize=12)

plt.tight_layout()

plt.savefig(
    '08_size_vs_installs.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
correlation = size_install_data['Size_MB'].corr(
    size_install_data['Installs']
)

print(
    "Correlation between App Size and Installs:",
    round(correlation, 3)
)
```

```python
size_install_data = size_install_data.copy()

size_install_data['Log_Installs'] = np.log10(
    size_install_data['Installs']
)

correlation = size_install_data['Size_MB'].corr(
    size_install_data['Log_Installs']
)

print(
    "Correlation between App Size and Log(Installs):",
    round(correlation, 3)
)
```
